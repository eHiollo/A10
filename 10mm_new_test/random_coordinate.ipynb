```python
import matplotlib.pyplot as plt
import math
import random
import numpy


# user defined input num 
num = 150
coordinates = []

while len(coordinates) < num:
        # 生成x坐标，范围在-0.003至+0.003之间，并保留五位小数
        x = round(random.uniform(-0.003, 0.003), 5)
        # 生成y坐标，范围在-0.003至+0.003之间，并保留五位小数
        y = round(random.uniform(-0.003, 0.003), 5)
        # 计算到原点的距离
        distance = math.sqrt(x**2 + y**2)
        # 如果距离在0.003以内，则保留该坐标
        if distance <= 0.003:
            coordinates.append((x, y))

print(f"{len(coordinates)}\t sets coordinates generate complete!")

with open('gen_coordinates_new_100.txt','w') as file:
    for coord in coordinates:
        file.write(f"{coord[0]}\t{coord[1]}\n")

print("Write Complete!")
```

```python
# Create figure and axis
fig, ax = plt.subplots(figsize=(8, 8))

# Draw the 10mm diameter circle (radius = 5mm)
outer_circle = plt.Circle((0, 0), 0.005, color='blue', fill=False, linewidth=2, label='10mm diameter')
ax.add_patch(outer_circle)

# Draw the 3mm diameter circle (radius = 3mm)
inner_circle = plt.Circle((0, 0), 0.003, color='red', fill=False, linewidth=2, linestyle='--', label='3mm diameter (±)')
ax.add_patch(inner_circle)

# Plot the random points
x_coords = [coord[0] for coord in coordinates]
y_coords = [coord[1] for coord in coordinates]
ax.scatter(x_coords, y_coords, color='green', s=20, label='Random points')

# Set plot limits and aspect ratio
ax.set_xlim(-0.0055, 0.0055)
ax.set_ylim(-0.0055, 0.0055)
ax.set_aspect('equal')

# Add labels and title
ax.set_xlabel('X position (meters)')
ax.set_ylabel('Y position (meters)')
ax.set_title('Random Points within 3mm Diameter Circle')
ax.legend()

# Show grid
ax.grid(True, linestyle='--', alpha=0.7)

plt.show()
```
